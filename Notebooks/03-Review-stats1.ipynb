## Recap: From Probability to Statistical Inference — Three Coin Tosses

### Experiment, sample space, and event

**Experiment:** Toss three fair coins.

The **sample space** is the set of all possible outcomes:

$$
S={HHH,HHT,HTH,HTT,THH,THT,TTH,TTT}.
$$

An **event** is a subset of the sample space. Let (A) be the event **“exactly two Heads are obtained.”**

$$
A={HHT,HTH,THH},
\qquad
P(A)=\frac{3}{8}.
$$

### Random variable and probability distribution

A **random variable** assigns a numerical value to each outcome. Let

$$
X=\text{number of Heads obtained in three coin tosses},
$$

so that

$$
X\in{0,1,2,3}.
$$

Its probability distribution is

| (x) | Outcomes      |  (P(X=x)) |
| --: | :------------ | :-------: |
|   0 | (TTT)         | ($\frac{1}{8}$) |
|   1 | (HTT,THT,TTH) | ($\frac{3}{8}$) |
|   2 | (HHT,HTH,THH) | ($\frac{3}{8}$) |
|   3 | (HHH)         | ($\frac{1}{8}$) |

Thus,

$$
X\sim\text{Binomial}(3,0.5).
$$

### Population and parameters

Imagine repeating this experiment a very large number of times. Each repetition produces a value ($X=0,1,2 \text{or} 3$).

The **population** is the collection of possible observations described by the probability model



A **parameter** is a numerical characteristic of this population. 
For example,$p=0.5$ is the probability of a Head, and the population mean is $\mu=E(X)=np=3(0.5)=1.5.
$

### Sample

Usually, we observe only a **sample** from the population. Suppose the experiment is repeated five times:

| Experiment | Outcome | (X_i) |
| ---------: | :-----: | ----: |
|          1 |  (HHT)  |     2 |
|          2 |  (HTT)  |     1 |
|          3 |  (HHH)  |     3 |
|          4 |  (THH)  |     2 |
|          5 |  (TTT)  |     0 |

Hence,

$$
X_1,\ldots,X_5={2,1,3,2,0}.
$$

### Sample statistics and estimates

A **statistic** is a numerical quantity calculated from a sample. 

The sample mean is$
\bar X=\frac{1}{n}\sum_{i=1}^{n}X_i
=\frac{2+1+3+2+0}{5}
=1.6.
$

Thus, ($\bar X=1.6$) is an **estimate** of the population mean ($\mu=1.5$).


### Putting the ideas together

| **Sample space**             | (${HHH,HHT,HTH,HTT,THH,THT,TTH,TTT}$)                     |

## Estimators and their properties

An **estimator** $\hat\theta$ is a rule (a function of the sample data) used to estimate an unknown population parameter $\theta$. Because it depends on a random sample, $\hat\theta$ is itself a random variable and has a **sampling distribution**. Its quality is judged by the location and spread of that distribution.

* **Bias:**
  $$
  \operatorname{Bias}(\hat\theta)=E(\hat\theta)-\theta.
  $$
  The estimator is **unbiased** if $E(\hat\theta)=\theta$, i.e. it does not systematically over- or under-estimate the parameter across repeated samples.

* **Efficiency:** among unbiased estimators, the one with the smaller variance is more precise, and is preferred.

* **Consistency:** as the sample size grows, the estimator concentrates on the truth, $\hat\theta\xrightarrow{p}\theta$.

* **Asymptotic unbiasedness:** any bias vanishes as $n\to\infty$, $\operatorname{Bias}(\hat\theta)\to 0$.

* **Asymptotic normality:** for many estimators,
  $$
  \sqrt{n}(\hat\theta-\theta)\xrightarrow{d}N(0,V),
  $$
  which justifies approximate confidence intervals and hypothesis tests in large samples.

* **Standard error:** the (estimated) standard deviation of $\hat\theta$'s sampling distribution. It measures sampling uncertainty and is the basis of confidence intervals and tests.

Popular methods for constructing estimators include **least squares**, **method of moments**, and **maximum likelihood**


| Method | Main idea |
|:---|:---|
| **Least squares (LS)** | Choose parameters that make fitted values close to observed values in squared-error terms |
| **Method of moments (MoM)** | Choose parameters so model-implied moments match sample moments |
| **Maximum likelihood (ML)** | Choose parameters under which the observed data are most compatible with the probability model |
